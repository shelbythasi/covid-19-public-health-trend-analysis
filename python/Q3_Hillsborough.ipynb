```python
import pandas as pd
import matplotlib.pyplot as plt

h_df = pd.read_excel('Combined Covid data set.xlsx', sheet_name='Hillsborough')
h_df.columns = ['Date', 'Daily Cases', 'Moving Average']
axes = h_df.plot('Date', ['Daily Cases'], xticks=[], color='orange')
h_df.plot('Date', ['Moving Average'], xticks=[], kind='bar', width=1, color='lightblue', ax=axes)

plt.xlabel('Date')
plt.ylabel('Cases')
plt.title('Hillsborough COVID-19')
plt.xticks(rotation=45)
plt.legend()
plt.show()
```
